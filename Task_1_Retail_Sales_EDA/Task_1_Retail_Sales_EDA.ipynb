```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales_dataset.csv")
```

```python
df.shape
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True, errors='coerce')
```

```python
df['order_date'].dtype
```

```python
df['order_date'].head()
```

```python
quarterly_sales = df.groupby(
    df['order_date'].dt.to_period('Q')
)['sales_amount'].sum()

quarterly_sales
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales (₹)')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
age_bins = [0, 18, 25, 35, 45, 55, 65, 100]
age_labels = ['<18', '18-24', '25-34', '35-44', '45-54', '55-64', '65+']

df['age_group'] = pd.cut(
    df['age'],
    bins=age_bins,
    labels=age_labels,
    right=False
)

age_group_counts = df['age_group'].value_counts().sort_index()

age_group_counts
```

```python
age_bins = [0, 18, 25, 35, 45, 55, 65, 100]
age_labels = ['<18', '18-24', '25-34', '35-44', '45-54', '55-64', '65+']

df['age_group'] = pd.cut(
    df['age'],
    bins=age_bins,
    labels=age_labels,
    right=False
)

age_group_counts = df['age_group'].value_counts().sort_index()

age_group_counts
```

```python
plt.figure(figsize=(10, 6))

age_group_counts.plot(kind='bar')

plt.title('Customer Age Group Distribution')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Observation

The customer base is concentrated mainly in the 25–34 and 35–44 age groups. The 25–34 group has the highest representation, followed by the 35–44 group. The 65+ group represents a very small portion of the dataset. The presence of unusually young customers should be reviewed during data-quality cleaning because the raw dataset contains invalid age values.

```python
top_products = (
    df.groupby('product_name')['quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
plt.figure(figsize=(12, 6))

top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Best-Selling Products')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Product')

plt.tight_layout()
plt.show()
```

### Observation

The chart identifies the top 10 products based on total quantity sold. These products have the highest sales volume and may represent important products for inventory planning and promotional campaigns. The business should ensure that high-demand products maintain sufficient inventory to avoid stockouts.

```python
category_revenue = (
    df.groupby('product_category')['sales_amount']
    .sum()
    .sort_values(ascending=False)
)

category_revenue
```

```python
plt.figure(figsize=(10, 6))

category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue (₹)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

Revenue varies across product categories, with some categories contributing substantially more sales than others. The highest-revenue categories should be monitored closely because they have a major impact on overall business revenue. Lower-performing categories may require pricing, promotion, or product-assortment strategies to improve their contribution.

```python
numeric_cols = df.select_dtypes(include='number')

correlation_matrix = numeric_cols.corr()

correlation_matrix
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Numerical Variables')

plt.tight_layout()
plt.show()
```

### Observation

The correlation matrix shows the relationships between numerical variables. Strong positive correlations indicate variables that tend to increase together, while negative correlations indicate an inverse relationship. The relationships between quantity, pricing, sales amount, profit, discount, and customer satisfaction can provide useful information for understanding the factors associated with sales performance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales_dataset.csv")

print(df.shape)
```

```python
region_profit = df.groupby('region')['profit'].sum()

print(region_profit)
```

```python
region_profit.plot(kind='bar')

plt.title('Total Profit by Region')
plt.xlabel('Region')
plt.ylabel('Total Profit (₹)')
plt.tight_layout()

plt.show()
```

### Observation

Profit contribution varies across the five regions. Some regions generate considerably higher total profit than others. This suggests that regional performance should be evaluated not only by sales volume but also by profitability. Regions with lower profit contribution could be investigated further by examining discounts, shipping costs, product categories, and sales performance.

```python
numeric_cols = df.select_dtypes(include='number')

descriptive_stats = pd.DataFrame({
    'Mean': numeric_cols.mean(),
    'Median': numeric_cols.median(),
    'Mode': numeric_cols.mode().iloc[0],
    'Standard Deviation': numeric_cols.std()
})

descriptive_stats
```

```python
print("Invalid ages:")
print(df[(df['age'] < 0) | (df['age'] > 100)][['order_id', 'age']])

print("\nInvalid quantities:")
print(df[df['quantity'] < 0][['order_id', 'quantity']])

print("\nNegative shipping costs:")
print(df[df['shipping_cost'] < 0][['order_id', 'shipping_cost']])

print("\nInvalid shipping days:")
print(df[df['days_to_ship'] < 0][['order_id', 'days_to_ship']])
```

### Data Quality Findings

The raw dataset contains several data-quality issues that require attention before using the data for business decisions. Customer age contains implausible values such as negative ages, 150, and 999. Quantity also contains negative values. The descriptive statistics additionally reveal negative shipping costs and negative shipping days, as well as unusually high values for quantity and age.

These anomalies should be investigated and corrected during the data-cleaning stage before performing further modelling or relying on the results for decision-making.

# Conclusion

This exploratory data analysis provided an overview of customer demographics, product performance, revenue distribution, regional profitability, and sales trends from 2020 to 2024.

### Key Business Insights

1. **Customer demographics:** The customer base is concentrated mainly in the 25–44 age range, making this an important customer segment for targeted marketing.

2. **Sales trends:** Quarterly sales fluctuate considerably across the study period. The highest observed quarterly sales occurred in 2023 Q1 at approximately ₹25.56 million.

3. **Product performance:** Rice (5kg), Jeans, Tennis Racket, Jacket, and Fiction Novel were among the highest-selling products by quantity in the raw dataset.

4. **Category performance:** Electronics generated the highest total sales among the product categories in the dataset.

5. **Regional profitability:** South generated the highest total profit among the five regions, while Central generated the lowest.

6. **Data quality:** The dataset contains missing values, duplicate records, and implausible values such as negative ages, negative quantities, and extreme age values. These issues should be addressed before using the data for production-level analysis.

### Business Recommendations

- Focus marketing campaigns on the major 25–44 customer segment.
- Maintain adequate inventory for high-volume products.
- Prioritize high-revenue categories such as Electronics while investigating opportunities in lower-performing categories.
- Investigate regional differences in profitability and identify ways to improve lower-performing regions.
- Clean and validate the dataset before using the results for important business decisions.
