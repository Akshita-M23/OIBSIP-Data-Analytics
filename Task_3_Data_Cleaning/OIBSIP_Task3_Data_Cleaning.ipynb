## Duplicate Removal

Duplicate rows were identified using pandas' `duplicated()` method and removed using `drop_duplicates()`. Removing exact duplicate records prevents the same transaction from being counted multiple times during analysis.

```python
empty_rows = df.isnull().all(axis=1).sum()

print("Completely empty rows:", empty_rows)
```

```python
df = df.dropna(how='all')
```

```python
df_raw = df.copy()

print("Original dataset copy created successfully!")
```

```python
missing_values = df.isnull().sum()

missing_values[missing_values > 0]
```

```python
numeric_columns = [
    'age',
    'quantity',
    'discount_pct',
    'customer_satisfaction',
    'days_to_ship'
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing numerical values handled.")
```

```python
df[numeric_columns].isnull().sum()
```

## Missing Numerical Values

Missing values in numerical columns were handled using median imputation. Median was selected because the dataset contains extreme values and is therefore less sensitive to outliers than the mean.

```python
categorical_columns = df.select_dtypes(
    include=['object', 'string']
).columns

print(df[categorical_columns].isnull().sum())
```

```python
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing categorical values handled.")
```

```python
df[categorical_columns].isnull().sum()
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
df['order_date'] = pd.to_datetime(
    df['order_date'],
    dayfirst=True,
    errors='coerce'
)
```

```python
print(df['order_date'].dtype)
print("Missing dates:", df['order_date'].isnull().sum())
```

```python
df['gender'] = df['gender'].astype(str).str.strip()

df['gender'] = df['gender'].replace({
    'M': 'Male',
    'm': 'Male',
    'MALE': 'Male',
    'male': 'Male',
    'F': 'Female',
    'f': 'Female',
    'FEMALE': 'Female',
    'female': 'Female'
})
```

```python
df['gender'].value_counts()
```

```python
df['order_status'] = (
    df['order_status']
    .astype(str)
    .str.strip()
    .str.title()
)

df['order_status'].value_counts()
```

```python
categorical_to_clean = [
    'region',
    'city',
    'product_category',
    'product_name',
    'payment_method'
]

for col in categorical_to_clean:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
df['order_id'] = df['order_id'].astype('string')
df['customer_id'] = df['customer_id'].astype('string')

numeric_columns_all = [
    'age',
    'quantity',
    'unit_price',
    'discount_pct',
    'sales_amount',
    'profit',
    'shipping_cost',
    'customer_satisfaction',
    'days_to_ship'
]

for col in numeric_columns_all:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.dtypes
```

```python
invalid_age = df[
    (df['age'] < 18) |
    (df['age'] > 100)
]

print("Invalid age records:", len(invalid_age))

invalid_age[['order_id', 'age']].head(20)
```

```python
df.loc[
    (df['age'] < 18) |
    (df['age'] > 100),
    'age'
] = np.nan

df['age'] = df['age'].fillna(df['age'].median())
```

```python
print("Minimum age:", df['age'].min())
print("Maximum age:", df['age'].max())
```

```python
invalid_quantity = df[df['quantity'] <= 0]

print("Invalid quantity records:", len(invalid_quantity))
```

```python
df.loc[df['quantity'] <= 0, 'quantity'] = np.nan

df['quantity'] = df['quantity'].fillna(
    df['quantity'].median()
)
```

```python
print("Minimum quantity:", df['quantity'].min())
```

```python
print(
    "Invalid shipping costs:",
    (df['shipping_cost'] < 0).sum()
)
```

```python
df.loc[
    df['shipping_cost'] < 0,
    'shipping_cost'
] = np.nan

df['shipping_cost'] = df['shipping_cost'].fillna(
    df['shipping_cost'].median()
)
```

```python
print(
    "Invalid shipping days:",
    (df['days_to_ship'] < 0).sum()
)
```

```python
df.loc[
    df['days_to_ship'] < 0,
    'days_to_ship'
] = np.nan

df['days_to_ship'] = df['days_to_ship'].fillna(
    df['days_to_ship'].median()
)
```

```python
invalid_discount = df[
    (df['discount_pct'] < 0) |
    (df['discount_pct'] > 1)
]

print("Invalid discount values:", len(invalid_discount))
```

```python
invalid_satisfaction = df[
    (df['customer_satisfaction'] < 1) |
    (df['customer_satisfaction'] > 5)
]

print(
    "Invalid satisfaction values:",
    len(invalid_satisfaction)
)
```

## Outlier Detection

The Interquartile Range (IQR) method was used to identify potential outliers in numerical variables.

Potential outliers were not automatically removed because extreme sales, profit, quantity, or price values may represent legitimate business transactions. Clearly implausible values, such as negative ages and negative shipping days, were treated separately as data-quality errors.

This approach helps preserve potentially meaningful business information while correcting values that are logically invalid.

```python
def find_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound
```

```python
columns_to_check = [
    'age',
    'quantity',
    'unit_price',
    'sales_amount',
    'profit',
    'shipping_cost',
    'days_to_ship'
]

for col in columns_to_check:
    outliers, lower, upper = find_outliers_iqr(df, col)

    print(
        col,
        "→ Outliers:", len(outliers),
        "| Lower:", round(lower, 2),
        "| Upper:", round(upper, 2)
    )
```

## Outlier Detection

The Interquartile Range (IQR) method was used to identify potential outliers in numerical variables.

```python
before_summary = pd.DataFrame({
    'Rows': [len(df_raw)],
    'Columns': [df_raw.shape[1]],
    'Duplicate Rows': [df_raw.duplicated().sum()],
    'Missing Values': [df_raw.isnull().sum().sum()]
})

after_summary = pd.DataFrame({
    'Rows': [len(df)],
    'Columns': [df.shape[1]],
    'Duplicate Rows': [df.duplicated().sum()],
    'Missing Values': [df.isnull().sum().sum()]
})

comparison = pd.concat(
    [before_summary, after_summary]
)

comparison.index = [
    'Before Cleaning',
    'After Cleaning'
]

comparison
```
